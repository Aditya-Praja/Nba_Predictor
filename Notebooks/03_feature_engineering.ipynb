```python
games["WL_home"] = games.apply(lambda x: 1 if x["WL_home"] == "W" else 0, axis=1)
games["WL_away"] = games.apply(lambda x: 1 if x["WL_away"] == "W" else 0, axis=1)
```

```python
games
```

```python
games["game_date"] = pd.to_datetime(games["game_date"])
```

```python
home = games[[
    "game_id", "game_date", "season",
    "home_team", "away_team",
    "home_win", "PTS_home", "PTS_away", "PLUS_MINUS_home"
]].copy()

home = home.rename(columns={
    "home_team": "team",
    "away_team": "opponent",
    "PTS_home": "points_for",
    "PTS_away": "points_against",
    "PLUS_MINUS_home": "point_diff",
})

home["is_home"] = 1
home["win"] = home["home_win"]

away = games[[
    "game_id", "game_date", "season",
    "away_team", "home_team",
    "home_win", "PTS_away", "PTS_home", "PLUS_MINUS_away"
]].copy()

away = away.rename(columns={
    "away_team": "team",
    "home_team": "opponent",
    "PTS_away": "points_for",
    "PTS_home": "points_against",
    "PLUS_MINUS_away": "point_diff",
})

away["is_home"] = 0
away["win"] = 1 - away["home_win"]

team_games = pd.concat([home, away], ignore_index=True)
team_games = team_games.sort_values(["team", "game_date", "game_id"])
```

```python
team_games.head()
```

```python
team_games
```

```python
team_games["games_played_before"] = team_games.groupby(["season", "team"]).cumcount()

team_games["season_wins_before"] = (
    team_games.groupby(["season", "team"])["win"]
    .cumsum()
    .shift(1)
)

team_games["season_win_pct_before"] = (
    team_games["season_wins_before"] / team_games["games_played_before"]
)
```

```python
team_games.head()
```

```python
team_games["last5_win_pct"] = (
    team_games.groupby(["season", "team"])["win"]
    .transform(lambda s: s.shift(1).rolling(5, min_periods=1).mean())
)

team_games["last5_point_diff"] = (
    team_games.groupby(["season", "team"])["point_diff"]
    .transform(lambda s: s.shift(1).rolling(5, min_periods=1).mean())
)
```

```python
team_games.head()
```

```python
team_games["last5_win_pct"] = (
    team_games.groupby(["season", "team"])["win"]
    .transform(lambda s: s.shift(1).rolling(5, min_periods=1).mean())
)

team_games["last5_point_diff"] = (
    team_games.groupby(["season", "team"])["point_diff"]
    .transform(lambda s: s.shift(1).rolling(5, min_periods=1).mean())
)

team_games["rest_days"] = (
    team_games.groupby("team")["game_date"]
    .diff()
    .dt.days
)
```

```python
team_games.head()
```

```python
feature_cols = [
    "game_id",
    "team",
    "season_win_pct_before",
    "last5_win_pct",
    "last5_point_diff",
    "rest_days",
]

home_features = team_games[team_games["is_home"] == 1][feature_cols].copy()
away_features = team_games[team_games["is_home"] == 0][feature_cols].copy()

home_features = home_features.add_prefix("home_")
away_features = away_features.add_prefix("away_")

features = games[[
    "game_id", "game_date", "season", "home_team", "away_team", "home_win"
]].copy()

features = features.merge(
    home_features,
    left_on=["game_id", "home_team"],
    right_on=["home_game_id", "home_team"],
    how="left",
)

features = features.merge(
    away_features,
    left_on=["game_id", "away_team"],
    right_on=["away_game_id", "away_team"],
    how="left",
)

features["win_pct_diff"] = (
    features["home_season_win_pct_before"]
    - features["away_season_win_pct_before"]
)

features["last5_win_pct_diff"] = (
    features["home_last5_win_pct"]
    - features["away_last5_win_pct"]
)

features["last5_point_diff_diff"] = (
    features["home_last5_point_diff"]
    - features["away_last5_point_diff"]
)

features["rest_days_diff"] = (
    features["home_rest_days"]
    - features["away_rest_days"]
)
```
