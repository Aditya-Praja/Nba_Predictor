```python
df["is_home"] = df["MATCHUP"].str.contains("vs.", regex=False)
```

```python
home = df[df["is_home"]][["GAME_ID", "TEAM_ABBREVIATION", "PTS"]].rename(
    columns={"TEAM_ABBREVIATION": "home_team", "PTS": "home_points"}
)

away = df[~df["is_home"]][["GAME_ID", "TEAM_ABBREVIATION", "PTS"]].rename(
    columns={"TEAM_ABBREVIATION": "away_team", "PTS": "away_points"}
)

game_df = home.merge(away, on="GAME_ID").rename(columns={"GAME_ID": "game_id"})
game_df = game_df[["game_id", "home_team", "away_team", "home_points", "away_points"]]
game_df.head()
```

```python
home = df[df["is_home"]].copy()
away = df[~df["is_home"]].copy()

home = home.set_index("GAME_ID")
away = away.set_index("GAME_ID")

home.columns = [f"{col}_home" for col in home.columns]
away.columns = [f"{col}_away" for col in away.columns]

game_df = home.merge(away, left_index=True, right_index=True).reset_index()
game_df.head()
```

```python
game_df["GAME_DATE"] = game_df["GAME_DATE_away"]
game_df["MATCHUP"] = game_df["MATCHUP_away"]
game_df["SEASON_ID"] = game_df["SEASON_ID_away"]
game_df["SEASON"] = game_df["SEASON_away"]
game_df.drop(columns=["is_home_home", "SEASON_ID_away", "SEASON_ID_home", "GAME_DATE_away", 
                      "GAME_DATE_home", "is_home_away", "SEASON_away", "SEASON_home", "MATCHUP_away", 
                      "MATCHUP_home", "VIDEO_AVAILABLE_away", "VIDEO_AVAILABLE_home"], inplace=True)
```
